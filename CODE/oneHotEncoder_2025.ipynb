### TRANFORMACION DE DATOS CATEGORICOS V.GENE.heavy, J.GENE.heavy, D.GENE.heavy, VJ.GENE.heavy

```python
from category_encoders import OneHotEncoder
import joblib
```

```python
# Columnas categóricas a codificar
columnas_categoricas = ['V.GENE.heavy', 'J.GENE.heavy', 'D.GENE.heavy', 'VJ.GENE.heavy']
```

```python
# Crear y aplicar OneHotEncoder de una sola vez
encoder = OneHotEncoder(cols=columnas_categoricas, use_cat_names=True)
df = encoder.fit_transform(df)
```

```python
# Guardar el encoder entrenado para luego usarlo con nuevos datos
joblib.dump(encoder, '../MODELS/onehot_encoder_v1_2025.sav')
```
